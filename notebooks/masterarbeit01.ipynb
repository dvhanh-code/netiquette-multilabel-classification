# Combine Unified Datasets

```python
import os
import pandas as pd

processed_dir = '/content/drive/MyDrive/netiquette_project/processed/'

# List all unified parquet files
unified_files = [f for f in os.listdir(processed_dir) if f.startswith('unified_') and f.endswith('.parquet')]

print(f"Found {len(unified_files)} unified parquet files:")
for f in unified_files:
    print(f)
```

```python
all_dfs = []

for file_name in unified_files:
    file_path = os.path.join(processed_dir, file_name)
    df = pd.read_parquet(file_path)
    all_dfs.append(df)

# Concatenate all DataFrames
merged_df = pd.concat(all_dfs, ignore_index=True)

print(f"Merged DataFrame shape: {merged_df.shape}")
display(merged_df.head())
```
