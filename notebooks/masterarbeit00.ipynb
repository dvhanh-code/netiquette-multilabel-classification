```python
import pandas as pd
import os

processed_data_dir = '/content/netiquette-multilabel-classification/data/processed/'

# Get a list of all unified parquet files
unified_files = [f for f in os.listdir(processed_data_dir) if f.startswith('unified_') and f.endswith('.parquet')]

# Read each file and store in a list of DataFrames
df_list = []
for file_name in unified_files:
    file_path = os.path.join(processed_data_dir, file_name)
    print(f"Read file: {file_name}")
    df_list.append(pd.read_parquet(file_path))

# Concatenate all DataFrames into one
combined_df = pd.concat(df_list, ignore_index=True)

print(f"Total data sentences after being combined: {len(combined_df)}")

# Display the first few rows of the combined DataFrame
display(combined_df.head())
```
